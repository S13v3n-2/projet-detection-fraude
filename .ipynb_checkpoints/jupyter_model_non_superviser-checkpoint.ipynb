```python
# Prétraitement
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler,LabelEncoder,OrdinalEncoder,OneHotEncoder
from datetime import datetime
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
```

```python
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])
df['dob'] = pd.to_datetime(df['dob'])
df['age_at_transaction'] = (df['trans_date_trans_time'] - df['dob']).dt.days / 365.25
df['age_at_transaction'] = df['age_at_transaction'].round(0).astype(int)
df['age_at_transaction']
```

```python
df.drop(["dob"], axis=1, inplace=True)
```

```python
df.sample(4)
```

```python
# S'assurer que la colonne est bien au format datetime
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])

# Extraire des caractéristiques numériques
df['hour'] = df['trans_date_trans_time'].dt.hour
df['day_of_week'] = df['trans_date_trans_time'].dt.dayofweek

# Supprimer la colonne originale qui pose problème
df = df.drop(columns=['trans_date_trans_time'])
```

```python
y = df["is_fraud"].astype(int)
X = df.drop(columns=["is_fraud"])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
numeric_features = ["amt", "zip", "city_pop", "age_at_transaction"]
#categorical_features_ord = ["Sex", "Embarked"]
categorical_features_non_ord = ["state", "city","gender","category","merchant","job",'last','first']

"""numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

"""categorical_ord_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])"""

categorical_non_ord_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ordinay", OrdinalEncoder())
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        #("cat_ord", categorical_ord_transformer, categorical_features_ord),
        ("cat_non_ord", categorical_non_ord_transformer, categorical_features_non_ord),
    ],
    remainder="drop"
)"""
```

```python
# Modèles de Machine Learning
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor,IsolationForest
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report, auc,roc_auc_score,precision_recall_curve
```

```python
model_IsoForest = IsolationForest(
    n_estimators=100,
    contamination=0.01, # On cible tes 1%
    max_samples='auto',
    random_state=42,
    n_jobs=-1
)

pipe = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("model", model_IsoForest)
])
```

```python
pipe.fit(X_train)
```

```python
y_pred = pipe.predict(X_test)
```

```python
# Obtenir les prédictions brutes (-1 et 1)
y_pred_raw = pipe.predict(X_test)

# Convertir : -1 devient 1 (Fraude), 1 devient 0 (Normal)
y_pred_converted = np.where(y_pred_raw == -1, 1, 0)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("--- ÉVALUATION ISOLATION FOREST ---")
print(classification_report(y_test, y_pred_converted, target_names=['Normal', 'Fraude']))

# Affichage de la matrice de confusion
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred_converted)
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds')
plt.xlabel('Prédiction (0=Normal, 1=Fraude)')
plt.ylabel('Réel')
plt.show()
```

```python
# Imaginons que y_pred_xgb sont les prédictions de ton XGBoost
# et y_pred_iso sont celles de ton Isolation Forest

fraud_found_by_iso = set(np.where(y_pred_converted == 1)[0])
fraud_found_by_xgb = set(np.where(y_pred_xgb == 1)[0])

# Les fraudes que SEUL l'Isolation Forest a vues :
new_discoveries = fraud_found_by_iso - fraud_found_by_xgb

print(f"Nombre de fraudes 'invisibles' débusquées par l'Isolation Forest : {len(new_discoveries)}")
```

```python
import numpy as np

# 1. On récupère les indices des fraudes pour chaque modèle
# On part du principe que tu as déjà y_pred_iso_converted et y_pred_xgb
fraud_found_by_iso = set(np.where(y_pred_iso_converted == 1)[0])
fraud_found_by_xgb = set(np.where(y_pred_xgb == 1)[0])

# 2. On crée la variable qui manquait : les découvertes uniques à l'Isolation Forest
new_discoveries_indices = fraud_found_by_iso - fraud_found_by_xgb

# 3. Vérification de sécurité pour y_test
# Si y_test est une Series Pandas, on la convertit en array pour l'indexation
y_true = y_test.values if hasattr(y_test, 'values') else np.array(y_test)

# 4. On filtre pour voir si ces découvertes sont de VRAIES fraudes
vraies_fraudes_debusquees = [idx for idx in new_discoveries_indices if y_true[idx] == 1]

# 5. Affichage des résultats
print(f"Nombre d'alertes uniques à l'Isolation Forest : {len(new_discoveries_indices)}")
print(f"Parmi elles, nombre de RÉELLES fraudes confirmées : {len(vraies_fraudes_debusquees)}")
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

# 1. EXTRAIRE les données transformées (numériques) via la pipeline
# On utilise la partie 'preprocess' de ta pipe pour transformer X_test
X_test_transformed = pipe.named_steps['preprocess'].transform(X_test)

# 2. Appliquer la PCA sur ces données transformées
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_test_transformed)

# 3. Récupérer les prédictions (là, ta pipe gère tout toute seule)
predictions = pipe.predict(X_test)

# 4. Affichage graphique
plt.figure(figsize=(10, 7))
sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=predictions,
    palette={1: 'dodgerblue', -1: 'red'},
    alpha=0.6,
    s=15
)

plt.title("Visualisation PCA : Répartition après Preprocessing")
plt.xlabel("Composante Principale 1")
plt.ylabel("Composante Principale 2")
# Correction de la légende pour correspondre aux couleurs
plt.legend(title="Prédiction", labels=['Anomalie (-1)', 'Normal (1)'])
plt.show()
```

```python
from sklearn.manifold import TSNE

# On prend un échantillon de 5000 points sinon c'est trop long
X_sample = X_test_transformed[:5000]
y_sample = predictions[:5000]

tsne = TSNE(n_components=2, perplexity=30, random_state=42)
X_tsne = tsne.fit_transform(X_sample)

plt.figure(figsize=(10, 7))
sns.scatterplot(x=X_tsne[:, 0], y=X_tsne[:, 1], hue=y_sample,
                palette={1: 'dodgerblue', -1: 'red'}, alpha=0.7)
plt.title("Visualisation t-SNE (Plus précise pour les structures locales)")
plt.show()
```

```python
# Score de décision : plus il est bas, plus c'est une anomalie
decision_scores = pipe.named_steps['model'].decision_function(X_test_transformed)

plt.figure(figsize=(10, 6))
sns.histplot(decision_scores, bins=50, kde=True)
plt.axvline(x=0, color='r', linestyle='--', label='Seuil 0')
plt.title("Séparation mathématique des données (Anomaly Scores)")
plt.xlabel("Score d'isolement")
plt.show()
```
